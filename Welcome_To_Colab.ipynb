```
for a in [True, False]:
    for b in [True, False]:
        print("Inside loop:",a,b)
        left = not (a and b)
        right = (not a) or (not b)
        print(a, b, "not(a and b):", left, "(not a) or (not b):", right, "match:", left == right)
```

```
spice_route = {"Anita", "Karthik"}
biryani_house = {"Ravi", "Divya"}
dosa_corner = {"Meena", "Anita", "Suresh"}
pizza_point = {"Ravi"}

print("Spice Route ∩ Dosa Corner:", spice_route & dosa_corner)
print("Spice Route ∪ Biryani House:", spice_route | biryani_house)
print("Biryani House - Pizza Point:", biryani_house - pizza_point)
print("Dosa Corner ∩ Biryani House:", dosa_corner & biryani_house)
```

```
order_customer_pairs = [
    (101, "Anita"), (102, "Ravi"), (103, "Meena"), (104, "Anita"),
    (105, "Karthik"), (106, "Divya"), (107, "Ravi"), (108, "Suresh"),
]

order_to_amount = {101: 540.0, 102: 380.0, 103: 260.0, 104: 120.0,
                    105: 910.0, 106: 450.0, 107: 690.0, 108: 300.0}
print("order_id -> amount is a function:",
      len(order_to_amount) == len(set(order_to_amount.keys())))

customer_to_orders = {}
for oid, cust in order_customer_pairs:
    customer_to_orders.setdefault(cust, []).append(oid)

multi_output = {c: o for c, o in customer_to_orders.items() if len(o) > 1}
print("customers with more than one order_id (breaks the function rule):", multi_output)
```
