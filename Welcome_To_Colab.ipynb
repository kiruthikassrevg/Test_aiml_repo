```
import sqlite3
conn = sqlite3.connect("orders.db")
cur = conn.cursor()
print("Database connected successfully!")
```

```
cur.execute("DROP TABLE IF EXISTS orders")
cur.execute("DROP TABLE IF EXISTS restaurants")
cur.execute("""
CREATE TABLE restaurants (
    restaurant_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    cuisine TEXT NOT NULL
)
""")
cur.execute("""

CREATE TABLE orders (
    order_id INTEGER PRIMARY KEY,
    customer TEXT NOT NULL,
    restaurant TEXT NOT NULL,
    items INTEGER NOT NULL,
    amount REAL NOT NULL,
    status TEXT NOT NULL
)
""")
conn.commit()
print("Tables created successfully!")
```

```
orders = [
    (101, "Anita", "Spice Route", 3, 540.0, "delivered"),
    (102, "Ravi", "Biryani House", 2, 380.0, "delivered"),
    (103, "Meena", "Dosa Corner", 4, 260.0, "cancelled"),
    (104, "Anita", "Dosa Corner", 1, 120.0, "delivered"),
    (105, "Karthik", "Spice Route", 5, 910.0, "delivered"),
    (106, "Divya", "Biryani House", 2, 450.0, "delivered"),
    (107, "Ravi", "Pizza Point", 3, 690.0, "cancelled"),
    (108, "Suresh", "Dosa Corner", 2, 300.0, "delivered")
]
restaurants = [
    (1, "Spice Route", "North Indian"),
    (2, "Biryani House", "Biryani"),
    (3, "Dosa Corner", "South Indian"),
    (4, "Pizza Point", "Italian")
]
cur.executemany(
    "INSERT INTO orders VALUES (?, ?, ?, ?, ?, ?)",
    orders
)
cur.executemany(
    "INSERT INTO restaurants VALUES (?, ?, ?)",
    restaurants
)
conn.commit()
print("Sample data inserted successfully!")
```

```
try:
    cur.execute("""
        INSERT INTO orders
        (order_id, customer, restaurant, items, status)
        VALUES (109, 'Priya', 'Dosa Corner', 2, 'delivered')
    """)
except sqlite3.IntegrityError as error:
    print("Database rejected the order:", error)
```

```
for row in cur.execute("""
    SELECT customer, amount
    FROM orders
"""):
    print(row)
```

```
for row in cur.execute("""
    SELECT order_id, customer, amount, status
    FROM orders
    WHERE amount >= 300
      AND status = 'delivered'
"""):
    print(row)
```

```
for row in cur.execute("""
    SELECT order_id, customer, amount, status
    FROM orders
    WHERE status = 'cancelled'
       OR amount > 800
"""):
    print(row)
```
