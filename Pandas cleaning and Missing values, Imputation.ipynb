```
import pandas as pd

sample = pd.DataFrame({
    "order_id": range(1, 11),
    "delivery_days": [2, 3, 1, 4, 999, 5, None, 999, 7, 3],
    "customer_rating": [4, 5, 2, 4, 3, 5, 4, 5, 4, 3],
    "delivery_zone": ["North", "North", "unknown", "North", None,
                      "South", "North", "South", "unknown", "North"],
})
sample.to_csv("delivery_data.csv", index=False)
```

```
df = pd.read_csv("delivery_data.csv")
print(df)
```

```
print(df.shape)
```

```
print(df.shape)              # (rows, columns)
print(df.columns.tolist())   # column names as a list
```

```
df.info()
print(df.describe())                    # non-null counts and data type per column
print(df.describe().T)
```

```
print(df.describe(include="all"))
print("-\n--")
print(df.describe(include="object"))  # count, unique, top, freq
print("---")
print(df.isnull().sum())              # missing values per column
print("---")
print(df.duplicated().sum())          # fully duplicated rows
print("---")
print(df.head())                      # first 5 rows
print(df.tail())                      # last rows
```

```
missing_markers = [999, "unknown"]

print(df["delivery_days"].unique())        # distinct values
print(df["delivery_days"].value_counts())  # how often each value occurs

df["delivery_days"] = df["delivery_days"].replace(missing_markers, pd.NA)
print(df)
print(df.isna().sum())
```

```
df["delivery_days"] = df["delivery_days"].replace(missing_markers, pd.NA)

df["delivery_zone"] = df["delivery_zone"].replace(missing_markers, pd.NA)
print(df["delivery_days"].isnull().sum())
```

Dropping, filling and taking a subset

```
cleaned = df.dropna()
print(cleaned.shape)
```

```
df["delivery_days"] = pd.to_numeric(df["delivery_days"])
df["delivery_days"] = df["delivery_days"].fillna(df["delivery_days"].mean())
print(df["delivery_days"].isnull().sum())  # 0
```

```
df["delivery_zone"] = df["delivery_zone"].fillna(df["delivery_zone"].mode()[0])
print(df["delivery_zone"].value_counts())
```

```
slow_deliveries = df[df["delivery_days"] > 5]
print(slow_deliveries)
```
