```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [200, 300, 450, 600, 800]
}

df = pd.DataFrame(data)

print(df)
```

```
sns.countplot(x="Month", data=df , color='yellow')
plt.bar(df["Month"], df["Sales"])
plt.xticks(rotation=90)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

Histogram example


```
plt.hist(df["Sales"], bins=8, edgecolor="black")

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Department": ["IT", "IT", "CSE", "CSE", "ECE", "ECE"],
    "Marks": [85, 90, 75, 80, 70, 95]
}

df = pd.DataFrame(data)

sns.barplot(data=df, x="Department", y="Marks")

plt.title("Department vs Average Marks")
plt.xlabel("Department")
plt.ylabel("Average Marks")

plt.show()
```
