```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [200, 300, 450, 600, 800]
}

df = pd.DataFrame(data)

print(df)
```

```
sns.countplot(x="Month", data=df , color='yellow')
plt.bar(df["Month"], df["Sales"])
plt.xticks(rotation=90)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

Histogram example


```
plt.hist(df["Sales"], bins=8, edgecolor="black")

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Department": ["IT", "IT", "CSE", "CSE", "ECE", "ECE"],
    "Marks": [85, 90, 75, 80, 70, 95]
}

df = pd.DataFrame(data)

sns.barplot(data=df, x="Department", y="Marks")

plt.title("Department vs Average Marks")
plt.xlabel("Department")
plt.xticks(rotation=30)
plt.ylabel("Average Marks")

plt.show()
```

```
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [200, 300, 450, 600, 800]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Jan", "Feb", "Mar"],
    "Product": ["Laptop", "Laptop", "Laptop",
                "Mobile", "Mobile", "Mobile"],
    "Sales": [200, 300, 400, 150, 250, 350]
}

df = pd.DataFrame(data)

sns.lineplot(
    data=df,
    x="Month",
    y="Sales",
    hue="Product",
    marker="o"
)

plt.title("Monthly Product Sales")
plt.show()
```
