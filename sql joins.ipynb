```
import sqlite3
conn = sqlite3.connect("orders.db")
cur = conn.cursor()
print("Database connected successfully!")
```

```
cur.execute("DROP TABLE IF EXISTS orders")
cur.execute("DROP TABLE IF EXISTS restaurants")
cur.execute("""
CREATE TABLE restaurants (
    restaurant_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    cuisine TEXT NOT NULL
)
""")
cur.execute("""

CREATE TABLE orders (
    order_id INTEGER PRIMARY KEY,
    customer TEXT NOT NULL,
    restaurant TEXT NOT NULL,
    items INTEGER NOT NULL,
    amount REAL NOT NULL,
    status TEXT NOT NULL
)
""")
conn.commit()
print("Tables created successfully!")
```

```
orders = [
    (101, "Anita", "Spice Route", 3, 540.0, "delivered"),
    (102, "Ravi", "Biryani House", 2, 380.0, "delivered"),
    (103, "Meena", "Dosa Corner", 4, 260.0, "cancelled"),
    (104, "Anita", "Dosa Corner", 1, 120.0, "delivered"),
    (105, "Karthik", "Spice Route", 5, 910.0, "delivered"),
    (106, "Divya", "Biryani House", 2, 450.0, "delivered"),
    (107, "Ravi", "Pizza Point", 3, 690.0, "cancelled"),
    (108, "Suresh", "Dosa Corner", 2, 300.0, "delivered")
]
restaurants = [
    (1, "Spice Route", "North Indian"),
    (2, "Biryani House", "Biryani"),
    (3, "Dosa Corner", "South Indian"),
    (4, "Pizza Point", "Italian")
]
cur.executemany(
    "INSERT INTO orders VALUES (?, ?, ?, ?, ?, ?)",
    orders
)
cur.executemany(
    "INSERT INTO restaurants VALUES (?, ?, ?)",
    restaurants
)
conn.commit()
print("Sample data inserted successfully!")
```

```
for row in cur.execute("""
    SELECT *
    FROM orders
"""):
    print(row)
for row in cur.execute("""
    SELECT *
    FROM restaurants
"""):
    print(row)
```

```
for row in cur.execute("""
    SELECT *
    FROM orders
"""):
    print(row)
for row in cur.execute("""
    SELECT *
    FROM restaurants
"""):
    print(row)
```

```
for row in cur.execute("""
    SELECT o.order_id,o.customer,o.restaurant,o.amount
    FROM orders o
    inner join restaurants r
    on o.restaurant=r.name
    order by o.customer;
"""):
    print(row)
```

```
for row in cur.execute("""
    SELECT o.order_id,o.customer,o.restaurant,o.amount,r.cuisine
    FROM orders o
    left join restaurants r
    on o.restaurant=r.name;
"""):
    print(row)
```

```
for row in cur.execute("""
    SELECT
        orders.order_id,
        orders.customer,
        orders.restaurant,
        restaurants.cuisine
    FROM orders
    LEFT JOIN restaurants
        ON orders.restaurant = restaurants.name
    ORDER BY orders.order_id
"""):
    print(row)
```

```
for row in cur.execute("""
    SELECT
        orders.order_id,
        orders.customer,
        orders.restaurant
    FROM orders
    LEFT JOIN restaurants
        ON orders.restaurant = restaurants.name
    WHERE restaurants.restaurant_id IS NULL
"""):
    print(row)
```

```
for row in cur.execute("""
    SELECT o.order_id,o.customer,o.restaurant,o.amount
    FROM orders o
    right join restaurants r
    on o.restaurant=r.name;
"""):
    print(row)
```

```
for row in cur.execute("""
    SELECT r.restaurant_id,r.cuisine,r.name
    FROM restaurants r
    right join orders o
    on r.name=o.restaurant;
"""):
    print(row)
```

```
for row in cur.execute("""
    SELECT
        restaurants.name,
        restaurants.cuisine,
        orders.order_id
    FROM orders
    RIGHT JOIN restaurants
        ON orders.restaurant = restaurants.name
    ORDER BY restaurants.restaurant_id, orders.order_id
"""):
    print(row)
```

```
for row in cur.execute("""
    SELECT
        orders.order_id,
        orders.customer,
        orders.restaurant,
        restaurants.cuisine
    FROM orders
    FULL OUTER JOIN restaurants
        ON orders.restaurant = restaurants.name
    ORDER BY orders.order_id
"""):
    print(row)

```
